the list.

```python
def sum1():
  list_of_num = [1,2,3,4,5,6]
  is_even = lambda x : x % 2 == 0
  total_sum = sum(list(filter(is_even , list_of_num)))
  return total_sum

sum1()
```

4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

for num in range(1, 201):
    if is_prime(num):
        print(f"{num} is prime")
    else:
        print(f"{num} is not prime")
```

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

```python
# prompt: Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

class Fibonacci:
  def __init__(self,n):
    self.n = n
    self.i = 0
    self.a = 0
    self.b =1

  def __iter__(self):
    return self

  def __next__(self):
        if self.i >= self.n:
            raise StopIteration


        self.a, self.b = self.b, self.a + self.b
        self.i += 1
        return self.a

for num in Fibonacci(5):
  print(num)
```

6 . Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def power_of_2(exponent):
  for i in range(exponent+1):
    yield 2**i

for power in power_of_2(5):
     print(power)
```

7.  Implement a generator function that reads a file line by line and yields each line as a string.

```python
import os

# Step 1: Create a file and write some content if it doesn't exist
filename = 'dict_content.txt'

# Check if the file already exists
if not os.path.exists(filename):
    with open(filename, 'w') as file:
        file.write("Line 1: This is a sample line.\n")
        file.write("Line 2: Another example line.\n")
        file.write("Line 3: Yet another line for testing.\n")

# Step 2: Generator function to read the file line by line
def read_file_line_by_line(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()  # Yield each line after stripping leading/trailing spaces

# Step 3: Using the generator to print each line
for line in read_file_line_by_line(filename):
    print(line)
```

9.Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
def celsius_to_Fehrenheit(c):
  return (c*9/5)+32

celsius_temp = [0,27,37,100]

fehrenheit_temp = list(map(celsius_to_Fehrenheit,celsius_temp))

print("celsius temp : ",celsius_temp)
print("fehrenheit temp : ",fehrenheit_temp)
```

10 .  Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
def is_not_vowels(char):
  return char.lower() not in 'aeiou'

input_string = "hello guys , python is amzing language"

filtered_char = filter(is_not_vowels,input_string)

join_string = ''.join(filtered_char)

print("original string : " , input_string)
print("filtered string : " , join_string )
```
